```python
def build_file_index(image_dir, mask_dir, file_extension='*.jpg',
                     mask_naming_pattern=None, mask_suffixes=None):
    """
    Build a list of {'img': Path, 'masks': [Path, ...]} dicts using pathlib.
    """
    image_dir = Path(image_dir)
    mask_dir = Path(mask_dir)

    # Path.glob returns Path objects; sorted() works directly on them
    # We sort by .name to replicate the old os.path.basename logic
    image_files = sorted(image_dir.glob(file_extension), key=lambda x: x.name)

    if not image_files:
        raise ValueError(f"No files matching {file_extension} in {image_dir}")

    index, missing = [], 0

    for img_path in image_files:
        base = img_path.stem  # Gets the filename without the extension
        mask_paths = []

        if mask_naming_pattern == 'suffix' and mask_suffixes:
            found_all = True
            for suffix in mask_suffixes:
                found_one = False
                # Check for various extensions using the / operator for joins
                for ext in ('.bmp', '.png', '.jpg', '.tif'):
                    candidate = mask_dir / f"{base}{suffix}{ext}"
                    if candidate.exists():
                        mask_paths.append(candidate)
                        found_one = True
                        break
                if not found_one:
                    print(f"  WARNING: mask not found for {base}{suffix}")
                    found_all = False
                    break
            if not found_all:
                mask_paths = [] # Reset if any suffix in the set is missing

        else:
            # Default: look for any image file with the same stem
            for ext in ('.png', '.jpg', '.jpeg', '.bmp', '.tif', '.tiff'):
                candidate = mask_dir / f"{base}{ext}"
                if candidate.exists():
                    mask_paths = [candidate]
                    break

        if mask_paths:
            index.append({'img': img_path, 'masks': mask_paths})
        else:
            missing += 1

    print(f"  Index built: {len(index)} valid pairs  |  {missing} skipped")
    if not index:
        raise ValueError(f"CRITICAL: No valid image-mask pairs found in {image_dir}!")

    return index
```

```python
import os
import shutil

base_label_dir = '/content/OCT500/Label'
output_masks_dir = os.path.join(base_label_dir, 'masks')

# Create the output masks directory if it doesn't exist
os.makedirs(output_masks_dir, exist_ok=True)
print(f"Created output directory: {output_masks_dir}")

# Define source folders (relative to base_label_dir) and their corresponding suffixes
# These are derived from the user's list of suffixes, assuming a 'GT_' prefix for folders.
# Based on previous output, folders like GT_Artery, GT_CAVF, GT_Capillary, GT_FAZ exist.
# Assuming GT_Vein and GT_LargeVessel also exist or will be skipped if not.

# CAPILLARY ARTERY VEIN FAZ
source_info = {
    'GT_FAZ': '_FAZ',
    'GT_Capillary': '_Capillary',
    'GT_Vein': '_Vein',
    'GT_Artery': '_Artery'
}

start_id = 10301 # Start processing images from this ID

print(f"Populating '{output_masks_dir}' with images from:")

for gt_folder_name, suffix in source_info.items():
    source_dir_path = os.path.join(base_label_dir, gt_folder_name)

    if not os.path.exists(source_dir_path):
        print(f"  Warning: Source directory '{source_dir_path}' does not exist. Skipping this mask type.")
        continue

    print(f"  - '{gt_folder_name}' (using suffix '{suffix}')")

    for filename in sorted(os.listdir(source_dir_path)):
        base_name, ext = os.path.splitext(filename)
        # Check if the base_name is numeric and greater than or equal to start_id
        if base_name.isdigit():
            img_id = int(base_name)
            if img_id >= start_id:
                # Construct the new filename: original_id + suffix + original_extension
                new_filename = f"{base_name}{suffix}{ext}"
                source_file = os.path.join(source_dir_path, filename)
                destination_file = os.path.join(output_masks_dir, new_filename)

                try:
                    shutil.copy2(source_file, destination_file)
                    # Optional: print for debugging specific copies
                    # print(f"    Copied '{filename}' to '{new_filename}'")
                except Exception as e:
                    print(f"    Error copying {source_file} to {destination_file}: {e}")
        # else:
            # print(f"    Skipping non-numeric filename: {filename}")

print(f"\nFinished populating the '{output_masks_dir}' directory.")

print(len(os.listdir("/content/OCT500/Label/masks")))
```

```python
import albumentations as A
```

```python
augmentors = [
    A.Compose([
        A.HorizontalFlip(p=1),
        A.VerticalFlip(p=0.5)
    ], additional_targets={'full_img': 'image', 'opl_bm_img': 'image'}),

    A.Compose([
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=1)
    ], additional_targets={'full_img': 'image', 'opl_bm_img': 'image'}),

    A.Compose([
        A.OneOf([
            A.GaussianBlur(blur_limit=(3, 5), p=0.5),
            A.GaussNoise(std_range=(0.1, 0.2), p=0.5),
        ], p=1)
    ], additional_targets={'full_img': 'image', 'opl_bm_img': 'image'}),

    A.Compose([
        A.ElasticTransform(alpha=1, sigma=50, p=1)
    ], additional_targets={'full_img': 'image', 'opl_bm_img': 'image'}),
]
```

```python
num = 10501
new_entries = []
new_labels  = []

for img_mask, img_label in zip(client_1_idx_octa, client_1_labels_octa):
    if img_label[1] == 1:
      continue
    old_id = img_mask['img'].stem  # e.g. '10449' — needed for mask renaming

    # ── Load the three projections ────────────────────────────────────────────
    # img_mask['img'] points to the ILM_OPL folder.
    # FULL and OPL_BM live in sibling folders with the same filename.
    parent_dir   = img_mask['img'].parents[1]  # two levels up from the file
    full_path    = parent_dir / 'OCTA(FULL)'   / img_mask['img'].name
    opl_bm_path  = parent_dir / 'OCTA(OPL_BM)' / img_mask['img'].name

    ilm_opl_img = cv2.cvtColor(cv2.imread(str(img_mask['img'])), cv2.COLOR_BGR2RGB)
    full_img    = cv2.cvtColor(cv2.imread(str(full_path)),        cv2.COLOR_BGR2RGB)
    opl_bm_img  = cv2.cvtColor(cv2.imread(str(opl_bm_path)),     cv2.COLOR_BGR2RGB)

    # ── Load masks ────────────────────────────────────────────────────────────
    masks = [cv2.imread(str(m), cv2.IMREAD_GRAYSCALE) for m in img_mask['masks']]

    # ── Apply each augmentor ──────────────────────────────────────────────────
    for i, aug in enumerate(augmentors):

        # Pass all three projections in one call.
        # Albumentations applies the SAME random transform to all of them.
        transformed = aug(
            image=ilm_opl_img,       # primary image → transformed['image']
            full_img=full_img,        # extra image   → transformed['full_img']
            opl_bm_img=opl_bm_img,   # extra image   → transformed['opl_bm_img']
            masks=masks               # masks         → transformed['masks']
        )

        aug_ilm_opl = transformed['image']
        aug_full    = transformed['full_img']
        aug_opl_bm  = transformed['opl_bm_img']
        aug_masks   = transformed['masks']

        new_id = str(num)

        # ── Save all three augmented projections to their own folders ─────────
        # Each projection goes into its own sibling directory, matching the
        # same folder structure as the original images.
        for aug_array, folder_name in [
            (aug_ilm_opl, 'OCTA(ILM_OPL)'),
            (aug_full,    'OCTA(FULL)'),
            (aug_opl_bm,  'OCTA(OPL_BM)'),
        ]:
            save_path = parent_dir / folder_name / f"{new_id}.bmp"
            cv2.imwrite(str(save_path), cv2.cvtColor(aug_array, cv2.COLOR_RGB2BGR))

        # ── Save augmented masks ──────────────────────────────────────────────
        new_mask_paths = []
        for orig_mask_path, aug_mask_array in zip(img_mask['masks'], aug_masks):
            new_mask_name = orig_mask_path.name.replace(old_id, new_id)
            new_mask_path = orig_mask_path.parent / new_mask_name
            cv2.imwrite(str(new_mask_path), aug_mask_array)
            new_mask_paths.append(new_mask_path)

        # ── The index entry still points to ILM_OPL as the primary image ─────
        # When load_single_sample runs, it will use img_mask['img'] to derive
        # the FULL and OPL_BM paths using the same parent_dir logic above.
        new_img_path = parent_dir / 'OCTA(ILM_OPL)' / f"{new_id}.bmp"
        new_entries.append({'img': new_img_path, 'masks': new_mask_paths})

        # Bug fix: use actual label, not hardcoded 0
        new_labels.append((new_id, img_label[1]))

        num += 1

client_1_idx_octa.extend(new_entries)
client_1_labels_octa.extend(new_labels)

print(f"Augmentation complete. Added {len(new_entries)} new samples.")
```

```python

# ─────────────────────────────────────────────────────────────────────────────
# Image utilities
# ─────────────────────────────────────────────────────────────────────────────

def Normalize_0_255(image):
    image = image - np.amin(image)
    if np.amax(image) > 0:
        image = image / np.amax(image)
    return (255 * image).astype(np.uint8)


def Image_intensity_equalize(image, clip_limit=1.0):
    try:
        if image is None or not isinstance(image, np.ndarray):
            raise TypeError("Input is not a valid image")
        image = (image - np.mean(image)) / (np.std(image) + 1e-8)
        image = Normalize_0_255(image)
        image = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
        planes = np.array(cv2.split(image))
        clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=(8, 8))
        planes[0] = clahe.apply(planes[0])
        image = cv2.merge(planes)
        image = cv2.cvtColor(image, cv2.COLOR_LAB2RGB)
        image = Normalize_0_255(image)
    except TypeError as e:
        print(f'Preprocessing error: {e}')
        return None
    return image


# ─────────────────────────────────────────────────────────────────────────────
# File index builder

# ─────────────────────────────────────────────────────────────────────────────
# Single-image loader
# ─────────────────────────────────────────────────────────────────────────────

def load_single_sample(img_path, mask_paths, img_size,
                       img_preprocess, output_channels,
                       mask_naming_pattern):
    """
    Load one (image, mask) pair and return float32 arrays normalised to [0,1].

    mask_naming_pattern determines how the mask files are interpreted:

    'suffix'
        Each mask file is a separate binary grayscale image representing one
        output channel.  This is now used for BOTH fundus (OC, OD) and OCTA
        (FAZ, CAVF, vessel layers).  Pixel values are normalised to [0,1].

    'gamma'
        A single grayscale image where pixel intensity encodes class membership:
            0   (black) = Optic Cup
            128 (gray)  = Optic Disc
            255 (white) = Background
        Two binary channels are created: OC channel and OD channel (which
        includes OC pixels since OD anatomically contains OC).
        Kept for legacy compatibility only.

    None / anything else
        Single grayscale mask, normalised to [0,1] as a single channel.
    """
    parent_dir = img_path.parents[1]
    if output_channels > 1: # octa
      full_img_path = parent_dir / 'OCTA(FULL)' / img_path.name
      opl_bm_img_path = parent_dir / 'OCTA(OPL_BM)' / img_path.name

      img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
      full_img = cv2.imread(full_img_path, cv2.IMREAD_GRAYSCALE)
      opl_bm_img = cv2.imread(opl_bm_img_path, cv2.IMREAD_GRAYSCALE)

      if img is None:
        raise ValueError(f"Failed to load image: {img_path}")
      if full_img is None:
        raise ValueError(f"Failed to load image: {full_img_path}")
      if opl_bm_img is None:
        raise ValueError(f"Failed to load image: {opl_bm_img_path}")

      img = cv2.resize(img, img_size)
      full_img = cv2.resize(full_img, img_size)
      opl_bm_img = cv2.resize(opl_bm_img, img_size)

      if img_preprocess:
        clahe = cv2.createCLAHE(clipLimit=1.0, tileGridSize=(8, 8))

        img = clahe.apply(img)
        full_img = clahe.apply(full_img)
        opl_bm_img = clahe.apply(opl_bm_img)

      img_arr = img.astype(np.float32) / 255.0
      full_img_arr = full_img.astype(np.float32) / 255.0
      opl_bm_img_arr = opl_bm_img.astype(np.float32) / 255.0

      img_arr = np.stack([img_arr, full_img_arr, opl_bm_img_arr], axis=-1)

    else:
      img = cv2.imread(img_path, cv2.IMREAD_COLOR)
      if img is None:
        raise ValueError(f"Failed to load image: {img_path}")

      img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
      img = cv2.resize(img, img_size)
      if img_preprocess:
        img = Image_intensity_equalize(img, clip_limit=1.0)
      img_arr = img.astype(np.float32) / 255.0





    # ── Mask loading ──────────────────────────────────────────────────────────
    if mask_naming_pattern == 'suffix':
        # One binary file per output channel — used for both fundus and OCTA
        channels = []
        for mp in mask_paths:
            m = cv2.imread(mp, cv2.IMREAD_GRAYSCALE)
            m = cv2.resize(
                m if m is not None else np.zeros(img_size, np.uint8),
                img_size)
            channels.append(m)
        mask_arr = np.stack(channels, axis=-1).astype(np.float32) / 255.0

    elif mask_naming_pattern == 'gamma':
        # Legacy: single image with semantic pixel intensities
        m = cv2.imread(mask_paths[0], cv2.IMREAD_GRAYSCALE)
        m = cv2.resize(
            m if m is not None else np.zeros(img_size, np.uint8),
            img_size)
        oc_mask  = (m == 0).astype(np.float32)    # black pixels = OC only
        od_mask  = (m <= 128).astype(np.float32)  # black + gray = OD (includes OC)
        mask_arr = np.stack([oc_mask, od_mask], axis=-1)

    else:
        m = cv2.imread(mask_paths[0], cv2.IMREAD_GRAYSCALE)
        m = cv2.resize(
            m if m is not None else np.zeros(img_size, np.uint8),
            img_size)
        mask_arr = m.astype(np.float32) / 255.0

    # ── Reconcile channel count ───────────────────────────────────────────────
    if mask_arr.ndim == 2:
        mask_arr = mask_arr[:, :, np.newaxis]

    actual_ch = mask_arr.shape[-1]
    if actual_ch < output_channels:
        pad = np.zeros((*img_size, output_channels - actual_ch), dtype=np.float32)
        mask_arr = np.concatenate([mask_arr, pad], axis=-1)
    elif actual_ch > output_channels:
        mask_arr = mask_arr[:, :, :output_channels]

    return img_arr, mask_arr


# ─────────────────────────────────────────────────────────────────────────────
# Label encoding helper
# ─────────────────────────────────────────────────────────────────────────────

def _get_onehot_label(img_path, label_dict, num_classes):
    """
    Look up the classification label for an image and return a one-hot vector.

    If the filename stem is not found in label_dict, defaults to class 0 and
    prints a warning.  This keeps training running if your label dict has minor
    gaps, but you should resolve any gaps before a real training run.

    Parameters
    ----------
    img_path    : full path to the image file
    label_dict  : {filename_stem: int_class_index}
    num_classes : total number of classes (for one-hot encoding)

    Returns
    -------
    one_hot : np.ndarray shape (num_classes,), dtype float32
    """
    stem      = os.path.splitext(os.path.basename(img_path))[0]
    cls_index = label_dict.get(stem, None)

    if cls_index is None:
        print(f"  WARNING: no label found for '{stem}', defaulting to class 0")
        cls_index = 0

    return np.eye(num_classes, dtype=np.float32)[int(cls_index)]


# ─────────────────────────────────────────────────────────────────────────────
# Balanced tf.data pipeline
# ─────────────────────────────────────────────────────────────────────────────

def build_balanced_tf_dataset(
    fundus_index, octa_index,
    img_size, img_preprocess,
    fundus_output_channels, octa_output_channels,
    fundus_mask_pattern, octa_mask_pattern,
    fundus_label_dict, octa_label_dict,
    num_fundus_classes, num_octa_classes,
    batch_size, shuffle,
):
    """
    Returns a balanced tf.data.Dataset (50% fundus, 50% OCTA per batch).

    Both generators now use identical one-hot encoding logic since both
    classification heads produce (num_classes,) softmax outputs.

    The inactive branch in each sample (e.g. OCTA fields in a fundus sample)
    receives zero arrays.  The masking logic in DualModalityModel._masked_loss()
    ensures these dummy zeros never contribute gradients during training.

    Parameters
    ----------
    fundus_label_dict  : {stem: int}  — 0=AMD, 1=DR, 2=Glaucoma, 3=Normal
    octa_label_dict    : {stem: int}  — 0=DR,  1=CNV, 2=Normal,  3=AMD
    num_fundus_classes : int          — 4
    num_octa_classes   : int          — 4
    """
    H, W = img_size

    print(f"\n  Creating pipeline:")
    print(f"    Fundus samples : {len(fundus_index)}")
    print(f"    OCTA samples   : {len(octa_index)}")

    # Handle single-modality datasets (one index is empty) —
    # used when building separate fundus-only or OCTA-only validation sets.
    if len(fundus_index) == 0 and len(octa_index) == 0:
        raise ValueError("Both fundus_index and octa_index are empty.")

    # ── Output signature — note both class label shapes are now (num_classes,) ─
    output_sig = (
        {
            'image_input':    tf.TensorSpec(shape=(H, W, 3), dtype=tf.float32),
            'modality_input': tf.TensorSpec(shape=(1,),      dtype=tf.float32),
        },
        {
            'fundus_output':       tf.TensorSpec(shape=(H, W, fundus_output_channels), dtype=tf.float32),
            'octa_output':         tf.TensorSpec(shape=(H, W, octa_output_channels),   dtype=tf.float32),
            # Both are now (num_classes,) one-hot vectors — no longer asymmetric
            'fundus_class_output': tf.TensorSpec(shape=(num_fundus_classes,), dtype=tf.float32),
            'octa_class_output':   tf.TensorSpec(shape=(num_octa_classes,),   dtype=tf.float32),
        }
    )

    # ── Fundus generator — class-balanced sampling ────────────────────────────
    # The core problem this solves: with ~2 fundus samples per batch and 4 classes,
    # Normal only appears in ~40% of batches by chance.  In all other batches it
    # receives small negative gradient updates from the softmax competition, which
    # accumulate and suppress it entirely over training.
    #
    # Solution: maintain a separate deque per class and yield from each class in
    # strict round-robin order.  This guarantees Normal appears every 4th sample,
    # so every batch of 2 fundus samples has a ~50% chance of containing Normal
    # (vs ~40% with random sampling) — and more importantly, Normal never goes
    # more than ~8 steps without a positive gradient update.
    def fundus_generator():
        from collections import defaultdict, deque

        # Group records by class label
        class_records = defaultdict(list)
        unclassified  = []
        for e in fundus_index:
            stem  = os.path.splitext(os.path.basename(e['img']))[0]
            label = fundus_label_dict.get(stem, None)
            bucket = int(label) if label is not None else -1
            rec = (e['img'], e['masks'], fundus_mask_pattern, fundus_output_channels)
            if bucket == -1:
                unclassified.append(rec)
            else:
                class_records[bucket].append(rec)

        num_cls    = num_fundus_classes
        class_keys = sorted(class_records.keys())

        # Build per-class deques; reshuffle and refill when exhausted
        def make_deque(records):
            lst = records.copy()
            if shuffle:
                np.random.shuffle(lst)
            return deque(lst)

        class_deques = {k: make_deque(class_records[k]) for k in class_keys}
        round_robin  = deque(class_keys)   # cycle through classes in order

        # Add any unlabelled samples to a fallback deque
        fallback = deque(unclassified)

        while True:
            # Pick the next class in round-robin order
            if round_robin:
                cls = round_robin[0]
                round_robin.rotate(-1)   # advance to next class

                # Refill this class's deque if exhausted (start of new epoch for it)
                if len(class_deques[cls]) == 0:
                    class_deques[cls] = make_deque(class_records[cls])

                img_path, mask_paths, mask_pattern, out_ch = class_deques[cls].popleft()
            else:
                # No class records at all — fall back to unclassified
                if len(fallback) == 0:
                    fallback = deque(unclassified)
                img_path, mask_paths, mask_pattern, out_ch = fallback.popleft()

            img_arr, mask_arr = load_single_sample(
                img_path, mask_paths, (W, H),
                img_preprocess, out_ch, mask_pattern)

            fundus_cls = _get_onehot_label(img_path, fundus_label_dict, num_fundus_classes)

            yield (
                {
                    'image_input':    img_arr.astype(np.float32),
                    'modality_input': np.array([0.0], dtype=np.float32),
                },
                {
                    'fundus_output':       mask_arr.astype(np.float32),
                    'octa_output':         np.zeros((H, W, octa_output_channels), dtype=np.float32),
                    'fundus_class_output': fundus_cls,
                    'octa_class_output':   np.zeros(num_octa_classes, dtype=np.float32),
                }
            )

    # ── OCTA generator — class-balanced round-robin + label filtering ─────────
    # Mirrors the fundus generator approach.  Samples whose label is not in
    # octa_label_dict are silently skipped (handles CNV/AMD removal cleanly:
    # just exclude them from octa_label_dict before passing it in, and they
    # will never appear in any batch).
    #
    # With only 29 DR vs 160 Normal, random sampling would suppress DR in the
    # same way Normal was suppressed for fundus.  Round-robin guarantees DR
    # appears every 2nd OCTA sample regardless of class frequency.
    def octa_generator():
        from collections import defaultdict, deque

        # Group by class — skip any entry whose stem is absent from the label dict
        class_records  = defaultdict(list)
        for e in octa_index:
            stem  = os.path.splitext(os.path.basename(e['img']))[0]
            label = octa_label_dict.get(stem, None)
            if label is None:
                continue                # silently skip filtered-out classes
            rec = (e['img'], e['masks'], octa_mask_pattern, octa_output_channels)
            class_records[int(label)].append(rec)

        class_keys = sorted(class_records.keys())
        if not class_keys:
            return   # nothing to yield (empty after filtering)

        def make_deque(records):
            lst = records.copy()
            if shuffle:
                np.random.shuffle(lst)
            return deque(lst)

        class_deques = {k: make_deque(class_records[k]) for k in class_keys}
        round_robin  = deque(class_keys)

        while True:
            cls = round_robin[0]
            round_robin.rotate(-1)

            if len(class_deques[cls]) == 0:
                class_deques[cls] = make_deque(class_records[cls])

            img_path, mask_paths, mask_pattern, out_ch = class_deques[cls].popleft()

            img_arr, mask_arr = load_single_sample(
                img_path, mask_paths, (W, H),
                img_preprocess, out_ch, mask_pattern)

            octa_cls = _get_onehot_label(img_path, octa_label_dict, num_octa_classes)

            yield (
                {
                    'image_input':    img_arr.astype(np.float32),
                    'modality_input': np.array([1.0], dtype=np.float32),
                },
                {
                    'fundus_output':       np.zeros((H, W, fundus_output_channels), dtype=np.float32),
                    'octa_output':         mask_arr.astype(np.float32),
                    'fundus_class_output': np.zeros(num_fundus_classes, dtype=np.float32),
                    'octa_class_output':   octa_cls,
                }
            )

    # ── Create dataset(s) and interleave if both modalities present ──────────
    total_samples = len(fundus_index) + len(octa_index)

    if len(fundus_index) == 0:
        # OCTA-only (e.g. OCTA-only validation set)
        octa_ds     = tf.data.Dataset.from_generator(octa_generator, output_signature=output_sig)
        combined_ds = octa_ds
    elif len(octa_index) == 0:
        # Fundus-only (e.g. fundus-only validation set)
        fundus_ds   = tf.data.Dataset.from_generator(fundus_generator, output_signature=output_sig)
        combined_ds = fundus_ds
    else:
        # Both modalities — interleave 50/50
        fundus_ds   = tf.data.Dataset.from_generator(fundus_generator, output_signature=output_sig)
        octa_ds     = tf.data.Dataset.from_generator(octa_generator,   output_signature=output_sig)
        combined_ds = tf.data.Dataset.sample_from_datasets(
            datasets=[fundus_ds, octa_ds],
            weights=[0.5, 0.5],
            stop_on_empty_dataset=False)

    combined_ds = (combined_ds
                   .batch(batch_size, drop_remainder=False)
                   .prefetch(tf.data.AUTOTUNE))

    return combined_ds, total_samples
```

```python
"""
Dual-modality LWBNA U-Net — segmentation + classification + Grad-CAM.

Architecture
------------
  Shared Encoder → Shared Bottleneck
    ├─ Fundus Decoder → [fundus_decoder_features] (B,320,320,128)
    │       ├─ Segmentation Head   → fundus_output        (B,320,320,1)  vascular mask (FIVES)
    │       └─ Classification Head → fundus_class_output  (B,4)          AMD/DR/Glaucoma/Normal
    └─ OCTA Decoder  → [octa_decoder_features]  (B,320,320,128)
            ├─ Segmentation Head   → octa_output          (B,320,320,6)  vessel masks
            └─ Classification Head → octa_class_output    (B,2)          Normal/DR (binary)

Both classification heads are now symmetric:
    GlobalAveragePooling2D → Dense(256, relu) → Dropout(0.5) → Dense(4, softmax)

This symmetry means _masked_loss() uses the same argmax-based accuracy
calculation for both branches, and both use categorical cross-entropy.

Joint loss
----------
  total = seg_weight × (fundus_dice_loss + octa_dice_loss)
        + cls_weight × (fundus_cce_loss  + octa_cce_loss)

  All per-sample losses are masked by modality so a fundus image never
  contributes to the OCTA loss and vice-versa.

Grad-CAM
--------
  build_grad_cam_models(model, branch) → (part1, part2)
  compute_grad_cam(part1, part2, image, modality_value, target_class, branch)
  visualise_grad_cam(model, image, ...)   ← convenience one-liner

  target_class is now meaningful for BOTH branches since both use softmax.
  For fundus: 0=AMD, 1=DR, 2=Glaucoma, 3=Normal  (or whatever order you define)
  For OCTA:   0=DR,  1=CNV, 2=Normal,  3=AMD      (or whatever order you define)
"""

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow import keras


# ─────────────────────────────────────────────────────────────────────────────
# Modality-connector layer
# ─────────────────────────────────────────────────────────────────────────────

class _ConnectModality(layers.Layer):
    """
    Adds zero contribution from modality_input to a segmentation output tensor.

    Purpose: Keras requires every Input() to be reachable from at least one
    model output.  modality_input is used inside train_step / test_step but is
    not naturally part of the segmentation forward pass.  This layer connects
    it to the graph by adding `reduce_mean(modality) * 0.0` — a value that is
    always exactly zero but keeps the computational graph connected so Keras
    does not raise an "unused Input" error.

    Why a custom layer instead of Lambda?
    Lambda layers in Keras 3 cannot automatically infer the output shape when
    their inputs have mismatched ranks (e.g. (B,320,320,1) + scalar).
    A custom layer with explicit compute_output_shape() avoids the
    NotImplementedError that Lambda raises in this situation.
    """

    def call(self, inputs):
        seg_out, modality = inputs
        # seg_out is already float32 (output Conv2D is marked dtype='float32').
        # modality_input is float32 from the generator.
        # No explicit casts needed — tf.reduce_mean preserves the input dtype.
        return seg_out + tf.reduce_mean(modality) * 0.0

    def compute_output_shape(self, input_shapes):
        # Output shape == segmentation tensor shape (first input)
        return input_shapes[0]


# ─────────────────────────────────────────────────────────────────────────────
# Loss / metric helpers
# ─────────────────────────────────────────────────────────────────────────────

def _per_sample_dice(gt, pred, smooth=1.0):
    """
    Dice coefficient computed independently for each sample in the batch.
    Returns shape (B,) — one scalar per image.
    We reduce over all spatial and channel dims [1,2,3] so multi-channel
    masks (e.g. OC + OD) are treated as a single joint segmentation.

    Both inputs are cast to float32 unconditionally.  Under mixed precision
    the model's internal activations are float16, and although the output
    Conv2D layers are marked dtype='float32', graph optimisations (XLA, the
    _ConnectModality wrapper, or kernel-fusion passes) can silently preserve
    float16 dtype on the tensor that actually arrives here.  Casting at the
    point of use is the only reliable defence — it is a no-op when the tensor
    is already float32, and costs negligible compute otherwise.
    """
    gt   = tf.cast(gt,   tf.float32)
    pred = tf.cast(pred, tf.float32)
    inter = tf.reduce_sum(gt * pred, axis=[1, 2, 3])
    union = tf.reduce_sum(gt,   axis=[1, 2, 3]) + \
            tf.reduce_sum(pred, axis=[1, 2, 3])
    return (2.0 * inter + smooth) / (union + smooth)

def dice_loss_binary(y_true, y_pred, smooth=1e-6):
    """Dice loss for a single-channel (binary) output."""
    # Flatten spatial dimensions
    y_true_f = tf.keras.backend.flatten(y_true)
    y_pred_f = tf.keras.backend.flatten(y_pred)

    intersection = tf.keras.backend.sum(y_true_f * y_pred_f)
    denominator = tf.keras.backend.sum(y_true_f) + tf.keras.backend.sum(y_pred_f)

    dice_score = (2. * intersection + smooth) / (denominator + smooth)
    return 1. - dice_score

def dice_loss_multichannel(y_true, y_pred, smooth=1e-6):
    """
    Dice loss for multi-channel independent outputs.
    Calculates Dice per channel, then averages the loss.
    y_true/y_pred shape: (batch, height, width, 4)
    """
    # Sum over spatial dimensions (height and width: axes 1 and 2)
    # This keeps the batch (axis 0) and channel (axis 3) separate
    intersection = tf.reduce_sum(y_true * y_pred, axis=[1, 2])
    denominator = tf.reduce_sum(y_true + y_pred, axis=[1, 2])

    dice_score = (2. * intersection + smooth) / (denominator + smooth)

    # Average the Dice score across all 4 channels and the batch
    return 1. - tf.reduce_mean(dice_score)

def dice_coef(y_true, y_pred, smooth=1.0):
    """Batch-averaged Dice coefficient (scalar). Useful as a standalone metric."""
    return tf.reduce_mean(_per_sample_dice(y_true, y_pred, smooth))


# ─────────────────────────────────────────────────────────────────────────────
# LWBNA building blocks
# ─────────────────────────────────────────────────────────────────────────────

def Attention(x, filter):
    """
    Channel-wise squeeze-and-excitation attention.
    Learns to re-weight feature channels based on their global average activation.
    """
    se = layers.GlobalAveragePooling2D()(x)
    se = layers.Dense(filter, kernel_initializer='he_normal')(se)
    se = layers.Activation('relu')(se)
    se = layers.Activation('sigmoid')(se)
    return layers.Multiply()([x, se])


def Convolution_block(x, filter=64, kernel=3, AB_layer=False, name_prefix=''):
    pfx = name_prefix + '_' if name_prefix else ''
    x = layers.Conv2D(filter, kernel, kernel_initializer='he_normal',
                      padding='same', name=f'{pfx}conv1' if pfx else None)(x)
    x = layers.BatchNormalization(name=f'{pfx}bn1'    if pfx else None)(x)
    x = layers.Activation('relu',     name=f'{pfx}relu1' if pfx else None)(x)
    x = layers.Conv2D(filter, kernel, kernel_initializer='he_normal',
                      padding='same', name=f'{pfx}conv2' if pfx else None)(x)
    x = layers.BatchNormalization(name=f'{pfx}bn2'    if pfx else None)(x)
    x = layers.Activation('relu',     name=f'{pfx}relu2' if pfx else None)(x)
    if AB_layer:
        x = Attention(x, filter)
    return x


def Encoder(x, depth_layer=4, f=16, dropout=0.3,
            AB_block=True, fixed_filter=False, name_prefix='enc'):
    skip_layers = []
    filt = f
    for i in range(depth_layer):
        filt = f if fixed_filter else f * (2 ** i)
        x = Convolution_block(x, filter=filt, kernel=3,
                              AB_layer=AB_block,
                              name_prefix=f'{name_prefix}_l{i}')
        skip_layers.append(x)
        x = layers.MaxPooling2D(2,  name=f'{name_prefix}_pool{i}')(x)
        x = layers.Dropout(dropout, name=f'{name_prefix}_drop{i}')(x)
    return x, skip_layers, filt


def Decoder(x, skip_layers, depth_layer=4, f=16,
            dropout=0.3, AB_block=True, name_prefix='dec'):
    for i in range(depth_layer):
        x = layers.UpSampling2D(2,  name=f'{name_prefix}_up{i}')(x)
        x = layers.add([x, skip_layers[depth_layer - 1 - i]],
                       name=f'{name_prefix}_add{i}')
        x = layers.Dropout(dropout, name=f'{name_prefix}_drop{i}')(x)
        x = Convolution_block(x, filter=f, kernel=3,
                              AB_layer=AB_block,
                              name_prefix=f'{name_prefix}_l{i}')
    return x


def Mid_block(x, f, depth_layer=4, r_factor=2,
              segmentation=False, name_prefix='mid'):
    f_red = 1
    xe1   = None
    out   = []
    for i in range(depth_layer):
        x = layers.Conv2D(int(f / f_red), 3, activation='relu',
                          kernel_initializer='he_normal',
                          padding='same', name=f'{name_prefix}_conv{i}')(x)
        if i == 0:
            xe1 = x
        x     = Attention(x, filter=int(f / f_red))
        f_red = f_red * r_factor
    out.append(x)
    if segmentation:
        x = layers.Conv2D(f, 3, activation='relu',
                          kernel_initializer='he_normal',
                          padding='same', name=f'{name_prefix}_conv_final')(x)
        x = layers.add([x, xe1], name=f'{name_prefix}_skip_add')
        out.append(x)
    return out


# ─────────────────────────────────────────────────────────────────────────────
# Custom model — joint segmentation + classification training
# ─────────────────────────────────────────────────────────────────────────────

class DualModalityModel(keras.Model):
    """
    Dual-modality model with joint segmentation + classification.

    Both branches are now fully symmetric in their classification heads:
    both use 4-class softmax + categorical cross-entropy.  This simplifies
    the loss function considerably compared to the previous binary/multiclass
    asymmetry.

    compile() accepts two extra kwargs:
        seg_weight  (float, default 1.0)  — weight on Dice losses
        cls_weight  (float, default 0.5)  — weight on CCE losses

    Start with cls_weight=0.5.  Increase if classification lags behind
    segmentation.  Decrease if segmentation quality degrades as classification
    improves, which would indicate the classification gradient is too strong.
    """

    def compile(self, optimizer, seg_weight=1.0,
                fundus_cls_weight=0.5, octa_cls_weight=0.3, **kwargs):
        """
        seg_weight        — multiplier on both Dice losses (default 1.0)
        fundus_cls_weight — multiplier on fundus CCE loss  (default 0.5)
        octa_cls_weight   — multiplier on OCTA CCE loss    (default 0.3)

        Gradient accumulation is handled in the custom training loop in
        data_pipeline.py, not here.  See Train_dual_modality_LWBNA_model
        for the accumulation_steps parameter.

        Separate per-branch classification weights let you keep fundus
        classification learning aggressively while damping the OCTA signal,
        which is noisier due to far fewer minority-class samples (29 DR vs
        160 Normal after filtering CNV and AMD out).
        """
        super().compile(optimizer=optimizer, **kwargs)
        self.seg_weight        = tf.constant(seg_weight,        dtype=tf.float32)
        self.fundus_cls_weight = tf.constant(fundus_cls_weight, dtype=tf.float32)
        self.octa_cls_weight   = tf.constant(octa_cls_weight,   dtype=tf.float32)
        # Pre-build optimiser slots eagerly so apply() is safe on first call
        self.optimizer.build(self.trainable_variables)

    def _masked_loss(self,
                     fundus_seg_gt,  fundus_seg_pred,
                     octa_seg_gt,    octa_seg_pred,
                     fundus_cls_gt,  fundus_cls_pred,
                     octa_cls_gt,    octa_cls_pred,
                     modality):
        """
        Compute per-modality masked losses and metrics.

        The masking logic is the key design decision here.  Because every
        batch contains a mix of fundus and OCTA images, and the inactive
        branch always receives dummy zero ground-truth labels from the
        generator, we must zero-out its contribution to the loss.

        For example, for a fundus sample (modality == 0.0):
          - is_fundus = 1.0  →  contributes to fundus_seg_loss and fundus_cls_loss
          - is_octa   = 0.0  →  its OCTA loss term is multiplied by 0, so the
                                 dummy zero OCTA masks/labels never affect training.

        Both branches now use categorical cross-entropy since both are 4-class
        softmax problems.  This is the main simplification from the previous
        binary fundus design.
        """
        # Modality masks — shape (B,)
        is_fundus = tf.cast(tf.equal(tf.squeeze(modality, 1), 0.0), tf.float32)
        is_octa   = tf.cast(tf.equal(tf.squeeze(modality, 1), 1.0), tf.float32)
        # Avoid division by zero when all samples in a batch are the same modality
        n_fundus  = tf.reduce_sum(is_fundus) + 1e-8
        n_octa    = tf.reduce_sum(is_octa)   + 1e-8

        # ── Segmentation: per-sample Dice, masked and averaged ────────────────
        # No manual float32 casts needed here — the output layers are already
        # marked dtype='float32' (the Conv2D seg heads and the Dense cls heads
        # via the _cls_cast linear activation), so predictions arrive as float32.
        # Ground truth from the generator is also float32 — types match.
        fundus_dice_ps = _per_sample_dice(fundus_seg_gt, fundus_seg_pred)  # (B,)
        num_octa_ch = octa_seg_gt.shape[-1]  # 3 (FAZ, Capillary, LargeVessel)

        octa_channel_dices = []
        octa_channel_bce   = []

        for c in range(num_octa_ch):
            gt_c   = octa_seg_gt[...,  c:c+1]   # (B, H, W, 1)
            pred_c = octa_seg_pred[..., c:c+1]  # (B, H, W, 1)

            bce_c = tf.keras.losses.binary_crossentropy(gt_c, pred_c)  # (B, H, W)
            bce_c = tf.reduce_mean(bce_c, axis=[1, 2])                 # (B,)
            octa_channel_bce.append(bce_c)

            ch_dice = _per_sample_dice(
                octa_seg_gt[..., c:c+1],    # shape (B, H, W, 1)
                octa_seg_pred[..., c:c+1])  # shape (B, H, W, 1)
            octa_channel_dices.append(ch_dice)  # each is shape (B,)

        # Stack and average across channels → still shape (B,)
        # Now FAZ, Capillary, and LargeVessel each contribute equally to the gradient,
        # regardless of how many pixels they occupy.
        octa_dice_ps = tf.reduce_mean(tf.stack(octa_channel_dices, axis=1), axis=1)
        octa_seg_bce  = tf.reduce_mean(tf.stack(octa_channel_bce,   axis=1), axis=1)

        fundus_seg_ce = tf.keras.losses.binary_crossentropy(
            fundus_seg_gt, fundus_seg_pred)
        fundus_seg_ce = tf.reduce_mean(fundus_seg_ce, axis=[1, 2])
        octa_seg_ce = tf.keras.losses.categorical_crossentropy(octa_seg_gt, octa_seg_pred)
        octa_seg_ce = tf.reduce_mean(octa_seg_ce, axis=[1, 2])

        fundus_seg_loss = tf.reduce_sum((fundus_seg_ce + (1.0 - fundus_dice_ps)) * is_fundus) / n_fundus
        octa_seg_loss = tf.reduce_sum((octa_seg_bce + (1.0 - octa_dice_ps)) * is_octa) / n_octa


        # fundus_seg_loss = tf.reduce_sum((1.0 - fundus_dice_ps) * is_fundus) / n_fundus
        # octa_seg_loss   = tf.reduce_sum((1.0 - octa_dice_ps)   * is_octa)   / n_octa

        # ── Classification: both branches now use categorical cross-entropy ───
        # Previously fundus used binary_crossentropy — that special case is gone.
        # Both branches are now identical in structure.
        fundus_cce = tf.keras.losses.categorical_crossentropy(
            fundus_cls_gt, tf.cast(fundus_cls_pred, tf.float32))             # (B,)
        octa_cce   = tf.keras.losses.categorical_crossentropy(
            octa_cls_gt,   tf.cast(octa_cls_pred,   tf.float32))             # (B,)

        fundus_cls_loss = tf.reduce_sum(fundus_cce * is_fundus) / n_fundus
        octa_cls_loss   = tf.reduce_sum(octa_cce   * is_octa)   / n_octa

        # ── Weighted total ────────────────────────────────────────────────────
        # Each classification branch has its own weight so the noisier OCTA
        # signal (few DR samples) can be damped independently of fundus.
        total = (self.seg_weight        * (fundus_seg_loss + 5*octa_seg_loss) +
                 self.fundus_cls_weight *  fundus_cls_loss +
                 self.octa_cls_weight   *  octa_cls_loss)

        # ── Metrics ───────────────────────────────────────────────────────────
        fundus_seg_dice = tf.reduce_sum(fundus_dice_ps * is_fundus) / n_fundus
        octa_seg_dice   = tf.reduce_sum(octa_dice_ps   * is_octa)   / n_octa

        # Accuracy: argmax comparison — now identical for both branches
        def _masked_accuracy(cls_gt, cls_pred, mask, n):
            pred_class = tf.argmax(cls_pred, axis=1, output_type=tf.int32)
            true_class = tf.argmax(cls_gt,   axis=1, output_type=tf.int32)
            correct    = tf.cast(tf.equal(pred_class, true_class), tf.float32)
            return tf.reduce_sum(correct * mask) / n

        fundus_cls_acc = _masked_accuracy(fundus_cls_gt, fundus_cls_pred, is_fundus, n_fundus)
        octa_cls_acc   = _masked_accuracy(octa_cls_gt,   octa_cls_pred,   is_octa,   n_octa)

        return (total,
                fundus_seg_loss, octa_seg_loss,
                fundus_cls_loss, octa_cls_loss,
                fundus_seg_dice, octa_seg_dice,
                fundus_cls_acc,  octa_cls_acc)

    def _unpack_batch(self, data):
        x, y = data
        return (x['image_input'], x['modality_input'],
                y['fundus_output'],       y['octa_output'],
                y['fundus_class_output'], y['octa_class_output'])

    @tf.function
    def train_step(self, data):
        """
        Single mini-batch forward + backward pass.  Returns gradients AND metrics
        so the custom training loop in data_pipeline.py can accumulate gradients
        across multiple mini-batches before applying the optimizer.

        @tf.function compiles this into a static graph so TF can reuse memory
        buffers and free intermediate activations immediately after use.
        Without it, eager mode keeps every activation alive until the full
        backward pass completes — causing OOM on large models even at batch_size=4.

        Gradient accumulation is intentionally NOT implemented here with
        tf.Variable accumulators.  Every attempt to use persistent tf.Variables
        inside train_step runs into Keras 3's XLA kernel-fusion, which fuses
        multiple train_step calls into a single GPU kernel during tracing.
        Variables created anywhere before that fused kernel (compile, lazy init,
        etc.) land on CPU, and XLA refuses cross-device access.

        Instead, accumulation is done in plain Python in the training loop:
            grads_list = [grad_batch_1, grad_batch_2, ...]
            averaged   = [sum(g) / N for g in zip(*grads_list)]
            optimizer.apply(averaged, model.trainable_variables)
        Python list operations happen outside tf.function scope, so there are
        no Variables and no device placement issues.
        """
        images, modality, f_seg_gt, o_seg_gt, f_cls_gt, o_cls_gt = \
            self._unpack_batch(data)

        with tf.GradientTape() as tape:
            f_seg_pred, o_seg_pred, f_cls_pred, o_cls_pred = self(
                {'image_input': images, 'modality_input': modality},
                training=True)
            (total, f_seg_loss, o_seg_loss, f_cls_loss, o_cls_loss,
             f_dice, o_dice, f_acc, o_acc) = self._masked_loss(
                f_seg_gt, f_seg_pred, o_seg_gt, o_seg_pred,
                f_cls_gt, f_cls_pred, o_cls_gt, o_cls_pred, modality)

        grads = tape.gradient(total, self.trainable_variables)
        grads = [g if g is not None else tf.zeros_like(v)
                 for g, v in zip(grads, self.trainable_variables)]

        # Return grads alongside metrics so the outer loop can accumulate them.
        # When called directly by model.fit() (no accumulation), the caller
        # should call optimizer.apply() itself — but train_step is now only
        # used by the custom loop, which handles that.
        return grads, {'loss': total,
                       'fundus_seg_loss': f_seg_loss, 'octa_seg_loss': o_seg_loss,
                       'fundus_cls_loss': f_cls_loss, 'octa_cls_loss': o_cls_loss,
                       'fundus_dice':     f_dice,      'octa_dice':     o_dice,
                       'fundus_cls_acc':  f_acc,        'octa_cls_acc':  o_acc}

    @tf.function
    def test_step(self, data):
        images, modality, f_seg_gt, o_seg_gt, f_cls_gt, o_cls_gt = \
            self._unpack_batch(data)

        f_seg_pred, o_seg_pred, f_cls_pred, o_cls_pred = self(
            {'image_input': images, 'modality_input': modality},
            training=False)
        (total, f_seg_loss, o_seg_loss, f_cls_loss, o_cls_loss,
         f_dice, o_dice, f_acc, o_acc) = self._masked_loss(
            f_seg_gt, f_seg_pred, o_seg_gt, o_seg_pred,
            f_cls_gt, f_cls_pred, o_cls_gt, o_cls_pred, modality)

        return {'loss': total,
                'fundus_seg_loss': f_seg_loss, 'octa_seg_loss': o_seg_loss,
                'fundus_cls_loss': f_cls_loss, 'octa_cls_loss': o_cls_loss,
                'fundus_dice':     f_dice,      'octa_dice':     o_dice,
                'fundus_cls_acc':  f_acc,        'octa_cls_acc':  o_acc}


# ─────────────────────────────────────────────────────────────────────────────
# Functional graph
# ─────────────────────────────────────────────────────────────────────────────

def _build_cls_head(features, num_classes, branch_name):
    """
    Shared classification head builder used by both branches.

    Both fundus and OCTA now use exactly the same head structure:
        GlobalAveragePooling2D → Dense(256, relu) → Dropout(0.5) → Dense(N, softmax)

    Extracting this into a helper makes the symmetry explicit and ensures
    the two branches are always kept in sync if you ever want to modify
    the head architecture.

    Parameters
    ----------
    features    : tensor (B, H, W, C) — the named decoder feature map
    num_classes : int — 4 for both branches in the current setup
    branch_name : 'fundus' or 'octa' — used to name the Keras layers

    Returns
    -------
    cls_output : tensor (B, num_classes) — softmax class probabilities
    """
    gap    = layers.GlobalAveragePooling2D(name=f'{branch_name}_gap')(features)
    d256   = layers.Dense(256, activation='relu',
                          name=f'{branch_name}_dense256')(gap)
    drop   = layers.Dropout(0.5, name=f'{branch_name}_cls_drop')(d256)
    # Cast to float32 before softmax — required for mixed precision stability.
    # float16 softmax can overflow/underflow; float32 is mandatory for output layers.
    drop_f32 = layers.Activation('linear', dtype='float32',
                                  name=f'{branch_name}_cls_cast')(drop)
    output = layers.Dense(num_classes, activation='softmax', dtype='float32',
                          name=f'{branch_name}_class_output')(drop_f32)
    return output


def build_dual_modality_graph(
    img_shape=(320, 320, 3),
    depth_layer=4,
    f=128,
    dropout=0.3,
    AB_block=True,
    fundus_output_channels=1,   # FIVES: single binary vascular mask
    octa_output_channels=6,
    num_fundus_classes=4,       # AMD, DR, Glaucoma, Normal
    num_octa_classes=2,         # Normal, DR  (CNV and AMD removed)
):
    """
    Build the functional Keras graph.

    Named layers used by Grad-CAM (do not rename these):
        '{branch}_decoder_features'  — (B, H, W, 128) spatial feature maps
        '{branch}_gap'               — GlobalAveragePooling2D
        '{branch}_dense256'          — Dense(256)
        '{branch}_class_output'      — Dense(N, softmax)
    """
    image_input    = keras.Input(shape=img_shape, name='image_input')
    modality_input = keras.Input(shape=(1,),       name='modality_input')

    # ── Shared encoder + bottleneck ───────────────────────────────────────────
    enc_out, skips, last_filt = Encoder(
        image_input, depth_layer=depth_layer, f=f,
        dropout=dropout, AB_block=AB_block,
        fixed_filter=True, name_prefix='shared_enc')

    bn = Mid_block(enc_out, last_filt, depth_layer=depth_layer,
                   r_factor=2, segmentation=True, name_prefix='shared_mid')
    bn = Convolution_block(bn[1], last_filt, kernel=3,
                           AB_layer=AB_block, name_prefix='shared_bn')

    # ── Fundus branch ─────────────────────────────────────────────────────────
    f_dec = Decoder(bn, skips, depth_layer=depth_layer, f=last_filt,
                    dropout=dropout, AB_block=AB_block, name_prefix='fundus_dec')

    # Named linear Activation = Grad-CAM hook point.
    # Activation('linear') is an identity op with an explicit name we can look
    # up later.  Avoids Lambda which has shape-inference issues in Keras 3.
    fundus_features = layers.Activation(
        'linear', name='fundus_decoder_features')(f_dec)

    # Segmentation head branches from the named feature map.
    # dtype='float32' is required for mixed precision — sigmoid in float16 is unstable.
    fundus_raw = layers.Conv2D(
        fundus_output_channels, 3,
        activation='sigmoid', padding='same',
        dtype='float32', name='fundus_raw')(fundus_features)

    # Classification head also branches from the same named feature map
    fundus_class = _build_cls_head(fundus_features, num_fundus_classes, 'fundus')

    # ── OCTA branch ───────────────────────────────────────────────────────────
    o_dec = Decoder(bn, skips, depth_layer=depth_layer, f=last_filt,
                    dropout=dropout, AB_block=AB_block, name_prefix='octa_dec')

    octa_features = layers.Activation(
        'linear', name='octa_decoder_features')(o_dec)

    octa_raw = layers.Conv2D(
        octa_output_channels, 3,
        activation='sigmoid', padding='same',
        dtype='float32', name='octa_raw')(octa_features)

    octa_class = _build_cls_head(octa_features, num_octa_classes, 'octa')

    # ── Wire modality_input into the graph ────────────────────────────────────
    # Keras requires every Input to connect to at least one output.
    # Multiplying by 0.0 makes modality_input a no-op on the actual values.
    # _ConnectModality links modality_input into the graph (required by Keras)
    # while contributing exactly zero to the output values.
    # Using a custom layer instead of Lambda avoids Keras 3's shape-inference
    # error when inputs have mismatched ranks.
    fundus_output = _ConnectModality(name='fundus_output')([fundus_raw,    modality_input])
    octa_output   = _ConnectModality(name='octa_output')(  [octa_raw,      modality_input])

    return (
        [image_input, modality_input],
        [fundus_output, octa_output, fundus_class, octa_class]
    )


# ─────────────────────────────────────────────────────────────────────────────
# Public factory
# ─────────────────────────────────────────────────────────────────────────────

def create_dual_modality_model(
    img_shape=(320, 320, 3),
    depth_layer=4,
    f=128,
    dropout=0.3,
    AB_block=True,
    fundus_output_channels=1,
    octa_output_channels=6,
    num_fundus_classes=4,
    num_octa_classes=2,
    use_custom_training=True,
):
    """
    Build and return the dual-modality LWBNA model.

    After calling this, compile with:
        model.compile(
            optimizer=keras.optimizers.Adam(1e-4),
            seg_weight=1.0,
            cls_weight=0.5,
        )
    """
    inputs, outputs = build_dual_modality_graph(
        img_shape=img_shape,
        depth_layer=depth_layer,
        f=f,
        dropout=dropout,
        AB_block=AB_block,
        fundus_output_channels=fundus_output_channels,
        octa_output_channels=octa_output_channels,
        num_fundus_classes=num_fundus_classes,
        num_octa_classes=num_octa_classes,
    )

    ModelClass = DualModalityModel if use_custom_training else keras.Model
    model = ModelClass(inputs=inputs, outputs=outputs, name='LWBNA_dual_modality')
    model.summary()
    return model


# ─────────────────────────────────────────────────────────────────────────────
# Grad-CAM
# ─────────────────────────────────────────────────────────────────────────────

def build_grad_cam_models(model, branch='fundus'):
    """
    Build the two sub-models needed for Grad-CAM on the given branch.

    Why split into two models?
    --------------------------
    GradientTape records operations as they happen.  To get gradients of the
    class score with respect to the spatial feature map, the tape needs to
    witness every operation between those two tensors.  If we run the full
    model first and then try to differentiate after the fact, the tape has no
    record of those intermediate operations — tape.gradient() returns None.

    The solution is to split the forward pass:
      part1: model.inputs  →  feature_map      (tape.watch() this output)
      part2: feature_map   →  class_scores     (tape records ops from here)

    Then inside compute_grad_cam():
        with tf.GradientTape() as tape:
            feat = part1(inputs)     # run part1 first
            tape.watch(feat)         # explicitly tell tape to watch this tensor
            scores = part2(feat)     # tape records all ops from feat to scores
        grads = tape.gradient(scores[0, target_class], feat)

    This works identically for both branches since they are now symmetric.

    Returns
    -------
    part1 : keras.Model  inputs       → feature_map (B, H, W, C)
    part2 : keras.Model  feature_map  → class probs  (B, num_classes)
    """
    assert branch in ('fundus', 'octa'), "branch must be 'fundus' or 'octa'"

    H, W       = model.input_shape[0][1], model.input_shape[0][2]
    feat_shape = model.get_layer(f'{branch}_decoder_features').output_shape
    feat_ch    = feat_shape[-1]

    # part1: full inputs → named feature map
    part1 = keras.Model(
        inputs=model.inputs,
        outputs=model.get_layer(f'{branch}_decoder_features').output,
        name=f'gradcam_part1_{branch}')

    # part2: feature map → class scores
    # Re-uses the EXACT same layer objects (and therefore weights) from the
    # trained model, so no weight copying is needed.
    feat_in = keras.Input(shape=(H, W, feat_ch), name='gradcam_feat_input')
    gap     = model.get_layer(f'{branch}_gap')(feat_in)
    d256    = model.get_layer(f'{branch}_dense256')(gap)
    # Dropout intentionally excluded — want deterministic heatmaps at inference
    cls_out = model.get_layer(f'{branch}_class_output')(d256)

    part2 = keras.Model(
        inputs=feat_in,
        outputs=cls_out,
        name=f'gradcam_part2_{branch}')

    return part1, part2


def compute_grad_cam(part1, part2, image, modality_value,
                     target_class=0, branch='fundus'):
    """
    Compute a Grad-CAM heatmap for a single image.

    Both branches now accept a meaningful target_class (0-3) since both
    use softmax.  You can call this once per class to see how the model's
    reasoning differs between, say, 'Glaucoma' and 'Normal' for the same image.

    Parameters
    ----------
    part1, part2   : from build_grad_cam_models()
    image          : np.ndarray (H, W, 3), float32 in [0, 1]
    modality_value : 0.0 = fundus, 1.0 = OCTA
    target_class   : int 0-3 — the class whose activation you want to explain
    branch         : 'fundus' or 'octa'

    Returns
    -------
    heatmap : np.ndarray (H, W), float32 in [0, 1]
    """
    img_batch = tf.cast(image[np.newaxis, ...], tf.float32)
    mod_batch = tf.constant([[modality_value]], dtype=tf.float32)

    with tf.GradientTape() as tape:
        feature_map  = part1([img_batch, mod_batch], training=False)
        tape.watch(feature_map)                         # watch AFTER part1 runs
        class_scores = part2(feature_map, training=False)
        loss         = class_scores[0, target_class]   # scalar — same for both branches

    grads = tape.gradient(loss, feature_map)            # (1, H, W, C)

    # Global-average the gradients over spatial dims to get per-channel weights
    pooled_grads = tf.reduce_mean(grads, axis=[0, 1, 2])   # (C,)

    # Weight the feature channels and collapse to a single spatial map
    feat_sq  = tf.squeeze(feature_map, axis=0)             # (H, W, C)
    heatmap  = feat_sq @ pooled_grads[..., tf.newaxis]     # (H, W, 1)
    heatmap  = tf.squeeze(heatmap).numpy()                 # (H, W)

    # ReLU: discard regions that suppress the target class
    heatmap = np.maximum(heatmap, 0)

    # Normalise to [0, 1]
    h_min, h_max = heatmap.min(), heatmap.max()
    heatmap = (heatmap - h_min) / (h_max - h_min + 1e-8)

    return heatmap.astype(np.float32)


def overlay_heatmap(image, heatmap, alpha=0.45, colormap=None):
    """Blend a Grad-CAM heatmap onto the original image. Returns uint8 RGB."""
    import cv2
    colormap = colormap or cv2.COLORMAP_JET
    img_u8   = (image * 255).astype(np.uint8) if image.max() <= 1.0 else image.astype(np.uint8)
    heat_rgb = cv2.cvtColor(
        cv2.applyColorMap((heatmap * 255).astype(np.uint8), colormap),
        cv2.COLOR_BGR2RGB)
    return cv2.addWeighted(img_u8, 1.0 - alpha, heat_rgb, alpha, 0)


def visualise_grad_cam(model, image, modality_value,
                       branch='fundus', target_class=0,
                       class_names=None, alpha=0.45):
    """
    One-call convenience: run inference + Grad-CAM + display.

    Example — explain why the model predicted 'Glaucoma' (index 2):
        visualise_grad_cam(model, img, modality_value=0.0, branch='fundus',
                           target_class=2,
                           class_names=['AMD', 'DR', 'Glaucoma', 'Normal'])

    Example — explain why the model predicted 'CNV' (index 1) for OCTA:
        visualise_grad_cam(model, img, modality_value=1.0, branch='octa',
                           target_class=1,
                           class_names=['DR', 'CNV', 'Normal', 'AMD'])
    """
    import matplotlib.pyplot as plt

    part1, part2 = build_grad_cam_models(model, branch=branch)

    img_batch = tf.cast(image[np.newaxis, ...], tf.float32)
    mod_batch = tf.constant([[modality_value]], dtype=tf.float32)
    preds = model({'image_input': img_batch, 'modality_input': mod_batch},
                  training=False)

    # Both branches are softmax — identical indexing
    cls_probs = preds[2 if branch == 'fundus' else 3].numpy()[0]  # (4,)
    pred_idx  = int(np.argmax(cls_probs))
    pred_name = class_names[pred_idx]   if class_names else f'Class {pred_idx}'
    expl_name = class_names[target_class] if class_names else f'Class {target_class}'
    pred_label = f'Predicted: {pred_name} ({cls_probs[pred_idx]:.3f})\nExplaining: {expl_name}'

    heatmap = compute_grad_cam(part1, part2, image, modality_value,
                               target_class=target_class, branch=branch)
    overlay = overlay_heatmap(image, heatmap, alpha=alpha)

    fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))
    axes[0].imshow(image);   axes[0].set_title(f'Input ({branch.upper()})'); axes[0].axis('off')
    im = axes[1].imshow(heatmap, cmap='jet', vmin=0, vmax=1)
    axes[1].set_title('Grad-CAM heatmap'); axes[1].axis('off')
    plt.colorbar(im, ax=axes[1], fraction=0.046, pad=0.04)
    axes[2].imshow(overlay); axes[2].set_title(pred_label); axes[2].axis('off')
    plt.tight_layout(); plt.show()

    return heatmap, overlay
```

```python
"""
Classification metrics callback for dual-modality training.

Computes proper epoch-level classification metrics on the full validation set:
    - Macro F1 score   (primary metric — use this for model selection)
    - Per-class F1     (diagnostic — tells you which classes are struggling)
    - Cohen's Kappa    (secondary metric — widely reported in ophthalmology)
    - Macro Precision  (supporting)
    - Macro Recall     (supporting)

Why not use accuracy?
----------------------
Accuracy is inappropriate for medical image classification because:
  1. Class imbalance is the norm (Normal cases dominate most datasets).
     A model that predicts "Normal" for everything scores well on accuracy
     while being clinically useless.  Macro F1 treats all classes equally
     regardless of how many samples each has.
  2. Batch-level accuracy is noisy — with batch_size=4 and 50/50 modality
     sampling you sometimes have a single fundus sample in a batch.
     Accuracy=1.0 on one correct sample is meaningless.  Epoch-level macro F1
     computed over the full validation set is stable and interpretable.

Accuracy is kept in train_step/test_step only as a fast rough proxy for
monitoring that training is progressing at all.  Use the metrics from this
callback for actual model evaluation and comparison.

Usage
-----
    from classification_metrics import ClassificationMetricsCallback

    callback = ClassificationMetricsCallback(
        validation_dataset  = valid_ds,
        validation_steps    = validation_steps,
        fundus_class_names  = ['AMD', 'DR', 'Glaucoma', 'Normal'],
        octa_class_names    = ['Normal', 'DR'],   # binary after CNV/AMD removal
        monitor_metric      = 'val_fundus_macro_f1',  # use for ModelCheckpoint
    )

    # Pass to model.fit() alongside your other callbacks.
    # IMPORTANT: If you use ModelCheckpoint with a metric from this callback,
    # pass this callback BEFORE ModelCheckpoint in the callbacks list so the
    # metric exists when ModelCheckpoint tries to read it.
    model.fit(
        train_ds,
        callbacks=[callback, checkpoint_cb, earlystop_cb],
        ...
    )

Metrics added to logs (accessible in history and other callbacks)
-----------------------------------------------------------------
    val_fundus_macro_f1         — macro-averaged F1 across all 4 fundus classes
    val_fundus_macro_precision  — macro precision
    val_fundus_macro_recall     — macro recall
    val_fundus_kappa            — Cohen's Kappa
    val_fundus_f1_<ClassName>   — per-class F1 for each fundus class
    val_octa_macro_f1           — macro F1 for OCTA
    val_octa_macro_precision    — macro precision
    val_octa_macro_recall       — macro recall
    val_octa_kappa              — Cohen's Kappa
    val_octa_f1_<ClassName>     — per-class F1 for each OCTA class
"""

import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.metrics import (
    f1_score, precision_score, recall_score,
    cohen_kappa_score, classification_report,
)


class ClassificationMetricsCallback(keras.callbacks.Callback):
    """
    Epoch-end callback that computes full-dataset classification metrics
    using sklearn on the complete validation set.

    Parameters
    ----------
    validation_dataset  : tf.data.Dataset — the validation dataset
    validation_steps    : int — number of batches to run (pass your
                          validation_steps from the training function)
    fundus_class_names  : list of str — e.g. ['AMD', 'DR', 'Glaucoma', 'Normal']
                          Order must match how you encoded labels (index 0 = first name)
    octa_class_names    : list of str — e.g. ['DR', 'CNV', 'Normal', 'AMD']
    monitor_metric      : str — which metric to print as the "headline" each epoch.
                          Defaults to 'val_fundus_macro_f1'.
    verbose             : bool — print a summary table each epoch
    print_every         : int — print the full classification report every N epochs.
                          The headline metric is always printed.
    """

    def __init__(self,
                 validation_dataset,
                 validation_steps,
                 fundus_class_names,
                 octa_class_names,
                 monitor_metric='val_fundus_macro_f1',
                 verbose=True,
                 print_every=10):
        super().__init__()
        self.val_ds           = validation_dataset
        self.val_steps        = validation_steps
        self.fundus_names     = fundus_class_names
        self.octa_names       = octa_class_names
        self.monitor_metric   = monitor_metric
        self.verbose          = verbose
        self.print_every      = print_every

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}

        # ── Collect predictions over the entire validation set ────────────────
        # We run the model manually here rather than relying on test_step
        # because we need the raw predicted class indices and true class indices
        # accumulated across all batches before we can compute F1/Kappa.
        fundus_true, fundus_pred = [], []
        octa_true,   octa_pred   = [], []

        step = 0
        for batch_x, batch_y in self.val_ds:
            if step >= self.val_steps:
                break

            modality = batch_x['modality_input'].numpy()  # (B, 1)

            # Run inference (training=False ensures dropout is off)
            outputs = self.model(batch_x, training=False)
            # outputs = [fundus_seg, octa_seg, fundus_cls, octa_cls]
            f_cls_probs = outputs[2].numpy()   # (B, num_fundus_classes)
            o_cls_probs = outputs[3].numpy()   # (B, num_octa_classes)

            f_cls_gt = batch_y['fundus_class_output'].numpy()  # (B, num_fundus_classes)
            o_cls_gt = batch_y['octa_class_output'].numpy()    # (B, num_octa_classes)

            # Split by modality — only accumulate active modality samples
            for i in range(modality.shape[0]):
                if modality[i, 0] == 0.0:   # fundus sample
                    fundus_true.append(np.argmax(f_cls_gt[i]))
                    fundus_pred.append(np.argmax(f_cls_probs[i]))
                else:                        # OCTA sample
                    octa_true.append(np.argmax(o_cls_gt[i]))
                    octa_pred.append(np.argmax(o_cls_probs[i]))

            step += 1

        # ── Compute and log metrics for each modality ─────────────────────────
        if fundus_true:
            f_metrics = self._compute_metrics(
                np.array(fundus_true), np.array(fundus_pred),
                self.fundus_names, 'fundus', logs, epoch)
        if octa_true:
            o_metrics = self._compute_metrics(
                np.array(octa_true), np.array(octa_pred),
                self.octa_names, 'octa', logs, epoch)

        # ── Print headline metric ─────────────────────────────────────────────
        if self.verbose and self.monitor_metric in logs:
            print(f"  {self.monitor_metric}: {logs[self.monitor_metric]:.4f}")

        f_f1 = logs.get('val_fundus_macro_f1', 0.0)
        o_f1 = logs.get('val_octa_macro_f1',   0.0)
        logs['val_mean_macro_f1'] = (f_f1 + o_f1) / 2.0

    def _compute_metrics(self, y_true, y_pred, class_names, branch, logs, epoch):
        """
        Compute macro F1, per-class F1, macro precision, macro recall, and
        Cohen's Kappa.  All results are written into `logs` so they appear in
        history and are visible to ModelCheckpoint / EarlyStopping.

        zero_division=0 means any class with no predicted samples gets F1=0
        rather than a warning.  This is the conservative, correct behaviour
        for medical imaging — a class the model never predicts is genuinely bad.
        """
        prefix = f'val_{branch}'
        # Explicit label list guarantees sklearn always returns num_classes scores,
        # even if some classes are absent from this validation split or never predicted.
        # Without this, sklearn sizes the output to however many unique classes appear
        # in y_true/y_pred — causing IndexError when a class is missing.
        all_labels = list(range(len(class_names)))

        macro_f1  = f1_score(y_true, y_pred, average='macro',     labels=all_labels, zero_division=0)
        macro_pre = precision_score(y_true, y_pred, average='macro', labels=all_labels, zero_division=0)
        macro_rec = recall_score(y_true, y_pred, average='macro',  labels=all_labels, zero_division=0)
        kappa     = cohen_kappa_score(y_true, y_pred, labels=all_labels)
        per_class_f1 = f1_score(y_true, y_pred, average=None, labels=all_labels, zero_division=0)  # always (num_classes,)

        # Write aggregate metrics into logs
        logs[f'{prefix}_macro_f1']        = float(macro_f1)
        logs[f'{prefix}_macro_precision']  = float(macro_pre)
        logs[f'{prefix}_macro_recall']     = float(macro_rec)
        logs[f'{prefix}_kappa']            = float(kappa)

        # Write per-class F1 into logs (e.g. 'val_fundus_f1_AMD')
        for cls_idx, cls_name in enumerate(class_names):
            safe_name = cls_name.replace(' ', '_')
            logs[f'{prefix}_f1_{safe_name}'] = float(per_class_f1[cls_idx])

        # ── Print detailed report every `print_every` epochs ─────────────────
        if self.verbose and (epoch + 1) % self.print_every == 0:
            n = len(y_true)
            print(f"\n  {'─'*50}")
            print(f"  {branch.upper()} Classification Report  (n={n} samples)")
            print(f"  {'─'*50}")
            print(classification_report(
                y_true, y_pred,
                labels=all_labels,
                target_names=class_names,
                zero_division=0,
                digits=3))
            print(f"  Macro F1 : {macro_f1:.4f}")
            print(f"  Kappa    : {kappa:.4f}")
            print(f"  {'─'*50}\n")

        return {
            'macro_f1':  macro_f1,
            'kappa':     kappa,
            'per_class': per_class_f1,
        }
```

```python

import time, math

# ─────────────────────────────────────────────────────────────────────────────
# FUNCTION 1: Pure training — one epoch, no validation, no callbacks
# ─────────────────────────────────────────────────────────────────────────────

def _train_one_epoch(model, train_iter, train_ds,
                     steps_per_epoch, accumulation_steps):
    """
    Run exactly one epoch of gradient-accumulation training.

    Completely separated from validation so it can be called by the FL
    client's fit() method without triggering any checkpointing or
    early stopping logic — those only make sense after evaluation.

    Parameters
    ----------
    model              : DualModalityModel (compiled, trainable vars set)
    train_iter         : iter(train_ds) — the CALLER owns this iterator and
                         passes it in. This allows the caller to maintain
                         a persistent iterator across multiple epoch calls,
                         which is important in FL where fit() is called
                         once per round rather than once per full training run.
    train_ds           : tf.data.Dataset — needed only to reset the iterator
                         if it is exhausted mid-epoch (edge case guard).
    steps_per_epoch    : int — mini-batches to consume this epoch
    accumulation_steps : int — mini-batches between optimizer updates

    Returns
    -------
    train_metrics : dict  {'loss': float, 'fundus_dice': float, ...}
                    Epoch-averaged training metrics. Does NOT have 'val_'
                    prefix — the caller adds those after evaluation.
    """
    updates_per_epoch = math.ceil(steps_per_epoch / accumulation_steps)
    train_totals = {}
    train_count  = 0
    mini_batch_idx = 0

    for update_idx in range(updates_per_epoch):
        accum_grads  = None
        step_metrics = {}

        for acc_step in range(accumulation_steps):
            if mini_batch_idx >= steps_per_epoch:
                break

            try:
                batch = next(train_iter)
            except StopIteration:
                # Infinite datasets should never exhaust, but guard anyway
                train_iter = iter(train_ds)
                batch = next(train_iter)

            grads, metrics = model.train_step(batch)

            if accum_grads is None:
                accum_grads = [g for g in grads]
            else:
                accum_grads = [ag + g for ag, g in zip(accum_grads, grads)]

            for k, v in metrics.items():
                step_metrics[k] = step_metrics.get(k, 0.0) + float(v)

            mini_batch_idx += 1

        n_acc     = acc_step + 1
        avg_grads = [g / tf.cast(n_acc, g.dtype) for g in accum_grads]
        model.optimizer.apply(avg_grads, model.trainable_variables)
        accum_grads = None
        avg_grads   = None

        for k, v in step_metrics.items():
            train_totals[k] = train_totals.get(k, 0.0) + v / float(n_acc)
        train_count += 1

        # Progress bar
        pct      = int(30 * (update_idx + 1) / updates_per_epoch)
        bar      = '=' * pct + '>' + '.' * (30 - pct)
        loss_now = step_metrics.get('loss', 0.0) / n_acc
        print(f"  [{bar}] {update_idx+1}/{updates_per_epoch}  "
              f"loss={loss_now:.4f}", end='\r')

    print()  # newline after progress bar
    return {k: v / max(train_count, 1) for k, v in train_totals.items()}


# ─────────────────────────────────────────────────────────────────────────────
# FUNCTION 2: Pure evaluation — validation + callbacks, no weight updates
# ─────────────────────────────────────────────────────────────────────────────

def _evaluate(model,
              fundus_val_ds, fundus_val_steps,
              octa_val_ds,   octa_val_steps,
              fundus_n, octa_n,
              callbacks, epoch,
              best_metric, epochs_no_impr):
    """
    Run validation, fire all callbacks, and return the complete logs dict.

    Completely separated from training so it can be called by the FL
    client's evaluate() method independently of fit().

    The caller owns `best_metric` and `epochs_no_impr` and passes them in.
    This is critical in FL: these dicts must persist across rounds on the
    client, not be reset each time evaluate() is called.

    Parameters
    ----------
    model              : DualModalityModel
    fundus_val_ds      : tf.data.Dataset — fundus-only validation set
    fundus_val_steps   : int
    octa_val_ds        : tf.data.Dataset — OCTA-only validation set
    octa_val_steps     : int
    fundus_n           : int — number of fundus validation samples
                         (used for weighted val_loss)
    octa_n             : int — number of OCTA validation samples
    callbacks          : list — same list passed to _custom_train_loop
    epoch              : int — 0-based epoch index (for callbacks)
    best_metric        : dict — mutable, owned by caller, updated in-place
    epochs_no_impr     : dict — mutable, owned by caller, updated in-place

    Returns
    -------
    val_logs     : dict  {'val_loss': float, 'val_fundus_dice': float, ...}
                   Contains ONLY validation keys — caller merges with
                   train_metrics to form the full epoch_logs.
    stop_training: bool  True if EarlyStopping fired this epoch.
    """

    # ── Separate callbacks by type ────────────────────────────────────────────
    cls_metric_cbs = [c for c in callbacks
                      if isinstance(c, ClassificationMetricsCallback)]
    checkpoint_cbs = [c for c in callbacks
                      if isinstance(c, keras.callbacks.ModelCheckpoint)]
    reduce_lr_cbs  = [c for c in callbacks
                      if isinstance(c, keras.callbacks.ReduceLROnPlateau)]
    early_stop_cbs = [c for c in callbacks
                      if isinstance(c, keras.callbacks.EarlyStopping)]

    # ── Run validation on each modality separately ────────────────────────────
    # Running separate loops avoids the dilution bug where fundus batches
    # contribute octa_dice=0.0 into the OCTA average.
    def _run_one_ds(ds, steps):
        totals = {}
        count  = 0
        for step, batch in enumerate(ds):
            if step >= steps:
                break
            metrics = model.test_step(batch)
            for k, v in metrics.items():
                totals[k] = totals.get(k, 0.0) + float(v)
            count += 1
        return {k: v / max(count, 1) for k, v in totals.items()}

    fundus_metrics = _run_one_ds(fundus_val_ds, fundus_val_steps)
    octa_metrics   = _run_one_ds(octa_val_ds,   octa_val_steps)

    # ── Build val_logs with clean, unambiguous keys ───────────────────────────
    # Each metric is averaged only over its own modality's batches.
    val_logs = {}

    # Weighted combined loss — the only metric that truly spans both modalities
    val_logs['val_loss'] = (
        fundus_n * fundus_metrics.get('loss', 0.0) +
        octa_n   * octa_metrics.get('loss',   0.0)
    ) / (fundus_n + octa_n)

    # Fundus-specific metrics — from fundus batches only
    val_logs['val_fundus_dice']    = fundus_metrics.get('fundus_dice',    0.0)
    val_logs['val_fundus_cls_acc'] = fundus_metrics.get('fundus_cls_acc', 0.0)
    val_logs['val_fundus_seg_loss']= fundus_metrics.get('fundus_seg_loss',0.0)
    val_logs['val_fundus_cls_loss']= fundus_metrics.get('fundus_cls_loss',0.0)

    # OCTA-specific metrics — from OCTA batches only
    val_logs['val_octa_dice']      = octa_metrics.get('octa_dice',        0.0)
    val_logs['val_octa_cls_acc']   = octa_metrics.get('octa_cls_acc',     0.0)
    val_logs['val_octa_seg_loss']  = octa_metrics.get('octa_seg_loss',    0.0)
    val_logs['val_octa_cls_loss']  = octa_metrics.get('octa_cls_loss',    0.0)

    fundus_dice_weight = 0.4
    octa_dice_weight   = 0.6
    val_logs['val_weighted_dice'] = (
        fundus_dice_weight * val_logs['val_fundus_dice'] +
        octa_dice_weight   * val_logs['val_octa_dice']
    )
    # ── Fire ClassificationMetricsCallback ────────────────────────────────────
    # This callback computes macro F1 / kappa and writes them into val_logs.
    # It must run BEFORE ModelCheckpoint so that val_mean_macro_f1 exists
    # when the checkpoint callback tries to read it.
    for cb in cls_metric_cbs:
        cb.on_epoch_end(epoch, logs=val_logs)

    # ── ReduceLROnPlateau ─────────────────────────────────────────────────────
    for cb in reduce_lr_cbs:
        if cb.monitor in val_logs:
            cb.on_epoch_end(epoch, logs=val_logs)

    # ── ModelCheckpoint ───────────────────────────────────────────────────────
    for cb in checkpoint_cbs:
        monitor  = cb.monitor
        mode     = getattr(cb, 'mode', None) or getattr(cb, '_mode', None)
        if mode not in ('min', 'max'):
            mode = 'min' if any(k in monitor for k in ('loss', 'error')) else 'max'
        val = val_logs.get(monitor, None)
        if val is None:
            continue
        prev_best = best_metric.get(monitor, None)
        improved  = (prev_best is None or
                     (mode == 'max' and val > prev_best) or
                     (mode == 'min' and val < prev_best))
        if improved:
            best_metric[monitor] = val
            model.save(cb.filepath)
            print(f"  ✓ Checkpoint saved  {monitor}={val:.4f}  → {cb.filepath}")

    # ── EarlyStopping ─────────────────────────────────────────────────────────
    stop_training = False
    for cb in early_stop_cbs:
        monitor  = cb.monitor
        mode     = getattr(cb, 'mode', None) or getattr(cb, '_mode', None)
        if mode not in ('min', 'max'):
            mode = 'min' if any(k in monitor for k in ('loss', 'error')) else 'max'
        min_d    = getattr(cb, 'min_delta', 1e-4)
        val      = val_logs.get(monitor, None)
        if val is None:
            continue
        prev_best = best_metric.get(f'es_{monitor}', None)
        if prev_best is None:
            best_metric[f'es_{monitor}'] = val
            epochs_no_impr[monitor]      = 0
        else:
            delta = val - prev_best if mode == 'max' else prev_best - val
            if delta > min_d:
                best_metric[f'es_{monitor}'] = val
                epochs_no_impr[monitor]      = 0
                if getattr(cb, 'restore_best_weights', False):
                    best_metric[f'es_weights_{monitor}'] = model.get_weights()
            else:
                epochs_no_impr[monitor] = epochs_no_impr.get(monitor, 0) + 1
                if epochs_no_impr[monitor] >= cb.patience:
                    print(f"\n  EarlyStopping: {monitor} did not improve "
                          f"for {cb.patience} epochs. Stopping.")
                    if getattr(cb, 'restore_best_weights', False):
                        saved = best_metric.get(f'es_weights_{monitor}')
                        if saved:
                            model.set_weights(saved)
                            print("  Best weights restored.")
                    stop_training = True

    return val_logs, stop_training

# ─────────────────────────────────────────────────────────────────────────────
# ORCHESTRATOR: thin loop that calls both functions in sequence
# ─────────────────────────────────────────────────────────────────────────────

def _custom_train_loop(model, train_ds,
                       fundus_val_ds, fundus_val_steps,
                       octa_val_ds,   octa_val_steps,
                       fundus_n, octa_n,
                       steps_per_epoch, validation_steps,
                       total_epochs, accumulation_steps, callbacks):

    class _History:
        def __init__(self): self.history = {}
        def _update(self, logs):
            for k, v in logs.items():
                self.history.setdefault(k, []).append(float(v))

    hist           = _History()
    best_metric    = {}
    epochs_no_impr = {}
    train_iter     = iter(train_ds)

    # Attach model to all callbacks once here, not inside the inner functions
    for cb in callbacks:
        cb.set_model(model)

    print(f"  Mini-batches/epoch : {steps_per_epoch}")
    print(f"  Optimizer updates  : {math.ceil(steps_per_epoch / accumulation_steps)}"
          f"  (every {accumulation_steps} mini-batches)")

    for epoch in range(total_epochs):
        epoch_start = time.time()
        print(f"\nEpoch {epoch + 1}/{total_epochs}")

        # ── Training ──────────────────────────────────────────────────────────
        train_metrics = _train_one_epoch(
            model, train_iter, train_ds,
            steps_per_epoch, accumulation_steps)

        # ── Validation + callbacks ────────────────────────────────────────────
        val_logs, stop_training = _evaluate(
            model,
            fundus_val_ds, fundus_val_steps,
            octa_val_ds,   octa_val_steps,
            fundus_n, octa_n,
            callbacks, epoch,
            best_metric, epochs_no_impr)

        # ── Merge and log ──────────────────────────────────────────────────────
        epoch_logs = {**train_metrics, **val_logs}

        elapsed      = time.time() - epoch_start
        summary_keys = ['loss', 'fundus_dice', 'octa_dice',
                        'val_loss', 'val_fundus_dice', 'val_octa_dice',
                        'val_fundus_macro_f1', 'val_octa_macro_f1']
        parts = [f"{k}={epoch_logs[k]:.4f}"
                 for k in summary_keys if k in epoch_logs]
        print(f"  {' — '.join(parts)}  ({elapsed:.0f}s)")

        hist._update(epoch_logs)

        if stop_training:
            break

    return hist


# def _custom_train_loop(model, train_ds, fundus_val_ds, fundus_val_steps, octa_val_ds,octa_val_steps,
#                        steps_per_epoch, validation_steps,
#                        total_epochs, accumulation_steps, callbacks):
#     """
#     Manual epoch/step loop that implements gradient accumulation in plain Python.

#     Why not model.fit() with a custom train_step?
#     -----------------------------------------------
#     Keras 3 wraps train_step in multi_step_on_iterator, which fuses N calls into
#     a single XLA-compiled GPU kernel.  Any tf.Variable created inside (or before)
#     that kernel — whether in compile(), lazily on first call, or anywhere else —
#     lands on CPU, and XLA refuses cross-device access with InvalidArgumentError.

#     The fix is to accumulate gradients as plain Python lists of tensors, never
#     as persistent tf.Variables.  Python list operations are outside tf.function
#     scope and have no device placement constraints.

#     Gradient accumulation mechanics
#     ---------------------------------
#     Each call to model.train_step() returns (grads, metrics) for one mini-batch
#     of batch_size images.  We sum those gradient tensors across accumulation_steps
#     mini-batches in a Python list, then divide by accumulation_steps to get the
#     average gradient, and call optimizer.apply() once.

#     This is mathematically identical to processing a single batch of
#     batch_size × accumulation_steps images — it is NOT an approximation.

#     Callbacks
#     ---------
#     We replicate the subset of Keras callback behaviour used by this project:
#         ModelCheckpoint  — saves when monitored metric improves
#         EarlyStopping    — stops when monitored metric stops improving
#         ReduceLROnPlateau — halves LR when val_loss plateaus
#         ClassificationMetricsCallback — calls on_epoch_end with logs dict

#     Parameters
#     ----------
#     model              : DualModalityModel instance (already compiled)
#     train_ds           : tf.data.Dataset (infinite, batched)
#     valid_ds           : tf.data.Dataset (finite, concatenated fundus+octa)
#     steps_per_epoch    : int — number of mini-batches per epoch
#     validation_steps   : int — number of batches to run for validation
#     total_epochs       : int
#     accumulation_steps : int — mini-batches between optimizer updates
#                          effective batch = batch_size × accumulation_steps
#     callbacks          : list — Keras callback objects (subset handled above)

#     Returns
#     -------
#     history_obj : object with .history dict  (mirrors keras History)
#     """
#     import time
#     import math

#     # ── History object ────────────────────────────────────────────────────────
#     class _History:
#         def __init__(self): self.history = {}
#         def _update(self, logs):
#             for k, v in logs.items():
#                 self.history.setdefault(k, []).append(float(v))

#     hist = _History()

#     # ── Callback helpers ──────────────────────────────────────────────────────
#     # Separate callbacks by type so we can call them at the right moments.

#     cls_metric_cbs = [c for c in callbacks
#                       if isinstance(c, ClassificationMetricsCallback)]
#     checkpoint_cbs = [c for c in callbacks
#                       if isinstance(c, keras.callbacks.ModelCheckpoint)]
#     reduce_lr_cbs  = [c for c in callbacks
#                       if isinstance(c, keras.callbacks.ReduceLROnPlateau)]
#     early_stop_cbs = [c for c in callbacks
#                       if isinstance(c, keras.callbacks.EarlyStopping)]

#     # Attach model to all callbacks (Keras normally does this in fit())
#     for cb in callbacks:
#         cb.set_model(model)

#     # ── State for EarlyStopping / ModelCheckpoint ─────────────────────────────
#     best_metric     = {}   # {monitor_key: best_value_seen}
#     epochs_no_impr  = {}   # {monitor_key: consecutive_epochs_without_improvement}
#     stop_training   = False

#     # ── Training iterator ─────────────────────────────────────────────────────
#     train_iter = iter(train_ds)

#     # Number of optimizer updates per epoch
#     # (each update covers accumulation_steps mini-batches)
#     updates_per_epoch = math.ceil(steps_per_epoch / accumulation_steps)

#     print(f"  Mini-batches/epoch : {steps_per_epoch}")
#     print(f"  Optimizer updates  : {updates_per_epoch}  "
#           f"(every {accumulation_steps} mini-batches)")

#     for epoch in range(total_epochs):
#         epoch_start = time.time()
#         print(f"\nEpoch {epoch + 1}/{total_epochs}")

#         # ── Training phase ────────────────────────────────────────────────────
#         # Running totals for epoch-level metric averages
#         train_totals = {}
#         train_count  = 0

#         mini_batch_idx = 0  # counts mini-batches within the epoch

#         for update_idx in range(updates_per_epoch):
#             # Accumulate gradients over accumulation_steps mini-batches
#             accum_grads   = None   # will be list of tensors, one per weight
#             step_metrics  = {}     # metric totals for this accumulation window

#             for acc_step in range(accumulation_steps):
#                 # Guard: don't exceed steps_per_epoch
#                 if mini_batch_idx >= steps_per_epoch:
#                     break

#                 try:
#                     batch = next(train_iter)
#                 except StopIteration:
#                     train_iter = iter(train_ds)
#                     batch = next(train_iter)

#                 # Forward + backward for this mini-batch
#                 grads, metrics = model.train_step(batch)

#                 # Accumulate gradients: initialise on first mini-batch,
#                 # then add to running sum on subsequent ones.
#                 if accum_grads is None:
#                     accum_grads = [g for g in grads]
#                 else:
#                     accum_grads = [ag + g for ag, g in zip(accum_grads, grads)]

#                 # Accumulate metrics (we'll average at the end of the window)
#                 for k, v in metrics.items():
#                     step_metrics[k] = step_metrics.get(k, 0.0) + float(v)

#                 mini_batch_idx += 1

#             n_acc = acc_step + 1  # actual number of mini-batches accumulated

#             # Average gradients and apply — wrapped in tf.function so the
#             # division and apply ops are graph-compiled, keeping memory usage
#             # equivalent to what model.fit() would produce.
#             avg_grads = [g / tf.cast(n_acc, g.dtype) for g in accum_grads]
#             model.optimizer.apply(avg_grads, model.trainable_variables)
#             # Release accumulator references immediately so GC can free GPU memory
#             accum_grads = None
#             avg_grads   = None

#             # Accumulate into epoch totals
#             for k, v in step_metrics.items():
#                 train_totals[k] = train_totals.get(k, 0.0) + v / float(n_acc)
#             train_count += 1

#             # Progress bar (simple)
#             pct = int(30 * (update_idx + 1) / updates_per_epoch)
#             bar = '=' * pct + '>' + '.' * (30 - pct)
#             loss_now = step_metrics.get('loss', 0) / n_acc
#             print(f"  [{bar}] {update_idx+1}/{updates_per_epoch}  "
#                   f"loss={loss_now:.4f}", end='\r')

#         # Epoch-average training metrics
#         epoch_logs = {k: v / max(train_count, 1) for k, v in train_totals.items()}
#         print()  # newline after progress bar

#         # # ── Validation phase ──────────────────────────────────────────────────
#         def run_validation(ds, steps):
#           totals = {}
#           count  = 0
#           for step, batch in enumerate(ds):
#               if step >= steps:
#                   break
#               metrics = model.test_step(batch)
#               for k, v in metrics.items():
#                   totals[k] = totals.get(k, 0.0) + float(v)
#               count += 1
#           return {k: v / max(count, 1) for k, v in totals.items()}

#         fundus_val_metrics = run_validation(fundus_val_ds, fundus_val_steps)
#         octa_val_metrics   = run_validation(octa_val_ds,   octa_val_steps)

#         # Merge into epoch_logs — each metric is averaged only over its own batches
#         for k, v in fundus_val_metrics.items():
#             epoch_logs[f'val_{k}'] = v   # fundus metrics from fundus batches only

#         for k, v in octa_val_metrics.items():
#             # For OCTA-specific metrics, overwrite with OCTA-only average.
#             # For shared metrics (like 'loss'), take a weighted average.
#             epoch_logs[f'val_octa_{k}'] = octa_val_metrics.get(k, 0)
#         n_f = len(fundus_idx_test)
#         n_o = len(octa_test_index)
#         epoch_logs['val_loss'] = (
#             n_f * fundus_val_metrics.get('loss', 0) +
#             n_o * octa_val_metrics.get('loss', 0)
#         ) / (n_f + n_o)

#         epoch_logs['val_fundus_dice'] = fundus_val_metrics.get('fundus_dice', 0)
#         epoch_logs['val_octa_dice']   = octa_val_metrics.get('octa_dice', 0)
#         epoch_logs['val_fundus_cls_acc'] = fundus_val_metrics.get('fundus_cls_acc', 0)
#         epoch_logs['val_octa_cls_acc']   = octa_val_metrics.get('octa_cls_acc',     0)
#         # val_totals = {}
#         # val_count  = 0
#         # for step, batch in enumerate(valid_ds):
#         #     if step >= validation_steps:
#         #         break
#         #     val_metrics = model.test_step(batch)
#         #     for k, v in val_metrics.items():
#         #         val_totals[k] = val_totals.get(k, 0.0) + float(v)
#         #     val_count += 1

#         # for k, v in val_totals.items():
#         #     epoch_logs[f'val_{k}'] = v / max(val_count, 1)

#         # ── Classification metrics callback ───────────────────────────────────
#         for cb in cls_metric_cbs:
#             cb.on_epoch_end(epoch, logs=epoch_logs)

#         # ── ReduceLROnPlateau ─────────────────────────────────────────────────
#         for cb in reduce_lr_cbs:
#             monitor = cb.monitor
#             if monitor in epoch_logs:
#                 cb.on_epoch_end(epoch, logs=epoch_logs)

#         # ── ModelCheckpoint ───────────────────────────────────────────────────
#         for cb in checkpoint_cbs:
#             monitor  = cb.monitor
#             # Keras 3 renamed/privatised .mode — read it defensively.
#             # Fall back to inferring from the monitor name: anything with
#             # 'loss' or 'error' in the name should be minimised; everything
#             # else (accuracy, f1, dice, kappa) should be maximised.
#             mode = getattr(cb, 'mode', None) or getattr(cb, '_mode', None)
#             if mode not in ('min', 'max'):
#                 mode = 'min' if any(k in monitor for k in ('loss', 'error')) else 'max'
#             val      = epoch_logs.get(monitor, None)
#             if val is None:
#                 continue
#             prev_best = best_metric.get(monitor, None)
#             improved  = (prev_best is None or
#                          (mode == 'max' and val > prev_best) or
#                          (mode == 'min' and val < prev_best))
#             if improved:
#                 best_metric[monitor] = val
#                 model.save(cb.filepath)
#                 print(f"  ✓ Checkpoint saved  {monitor}={val:.4f}  → {cb.filepath}")

#         # ── EarlyStopping ─────────────────────────────────────────────────────
#         for cb in early_stop_cbs:
#             monitor  = cb.monitor
#             mode = getattr(cb, 'mode', None) or getattr(cb, '_mode', None)
#             if mode not in ('min', 'max'):
#                 mode = 'min' if any(k in monitor for k in ('loss', 'error')) else 'max'
#             min_d    = getattr(cb, 'min_delta', 1e-4)
#             val      = epoch_logs.get(monitor, None)
#             if val is None:
#                 continue
#             prev_best = best_metric.get(f'es_{monitor}', None)
#             if prev_best is None:
#                 best_metric[f'es_{monitor}'] = val
#                 epochs_no_impr[monitor] = 0
#             else:
#                 delta = val - prev_best if mode == 'max' else prev_best - val
#                 if delta > min_d:
#                     best_metric[f'es_{monitor}'] = val
#                     epochs_no_impr[monitor] = 0
#                     if getattr(cb, 'restore_best_weights', False):
#                         # Save current weights as best
#                         best_metric[f'es_weights_{monitor}'] = model.get_weights()
#                 else:
#                     epochs_no_impr[monitor] = epochs_no_impr.get(monitor, 0) + 1
#                     if epochs_no_impr[monitor] >= cb.patience:
#                         print(f"\n  EarlyStopping: {monitor} did not improve "
#                               f"for {cb.patience} epochs. Stopping.")
#                         if getattr(cb, 'restore_best_weights', False):
#                             saved = best_metric.get(f'es_weights_{monitor}')
#                             if saved:
#                                 model.set_weights(saved)
#                                 print("  Best weights restored.")
#                         stop_training = True

#         # ── Print epoch summary ───────────────────────────────────────────────
#         elapsed = time.time() - epoch_start
#         summary_keys = ['loss', 'fundus_dice', 'octa_dice',
#                         'val_loss', 'val_fundus_dice', 'val_octa_dice',
#                         'val_fundus_macro_f1', 'val_octa_macro_f1']
#         parts = [f"{k}={epoch_logs[k]:.4f}" for k in summary_keys if k in epoch_logs]
#         print(f"  {' — '.join(parts)}  ({elapsed:.0f}s)")

#         hist._update(epoch_logs)

#         if stop_training:
#             break

#     return hist


# ─────────────────────────────────────────────────────────────────────────────
# Training curve plotting
# ─────────────────────────────────────────────────────────────────────────────

def _plot_training_curves(history):
    h = history.history
    fig, axes = plt.subplots(2, 2, figsize=(16, 10))
    axes = axes.flatten()

    # Total loss
    axes[0].plot(h.get('loss', []),     label='train')
    axes[0].plot(h.get('val_loss', []), label='val')
    axes[0].set_title('Total Loss'); axes[0].set_xlabel('Epoch')
    axes[0].legend(); axes[0].grid(True)

    # Segmentation Dice
    for key, lbl in [('fundus_dice',     'fundus train'), ('val_fundus_dice', 'fundus val'),
                      ('octa_dice',       'octa train'),   ('val_octa_dice',   'octa val')]:
        if key in h:
            axes[1].plot(h[key], label=lbl)
    axes[1].set_title('Segmentation Dice'); axes[1].set_xlabel('Epoch')
    axes[1].legend(); axes[1].grid(True)

    # Macro F1 (primary classification metric)
    for key, lbl in [('val_fundus_macro_f1', 'fundus val F1'),
                      ('val_octa_macro_f1',   'octa val F1')]:
        if key in h:
            axes[2].plot(h[key], label=lbl)
    axes[2].set_title('Macro F1 Score (val)'); axes[2].set_xlabel('Epoch')
    axes[2].set_ylim([0, 1]); axes[2].legend(); axes[2].grid(True)

    # Cohen's Kappa
    for key, lbl in [('val_fundus_kappa', 'fundus'),
                      ('val_octa_kappa',   'octa')]:
        if key in h:
            axes[3].plot(h[key], label=lbl)
    axes[3].set_title("Cohen's Kappa (val)"); axes[3].set_xlabel('Epoch')
    axes[3].set_ylim([-0.1, 1]); axes[3].legend(); axes[3].grid(True)

    plt.tight_layout()
    plt.show()
    plt.close(fig)
```

```python
class Client(fl.client.NumPyClient):
    def __init__(self, client_id,
                 # Pass lightweight data references, not datasets
                 octa_train_index, octa_label_dict_train,
                 fundus_idx_train, fives_label_dict_train,
                 # Hyperparameters — all plain Python, safe to pickle
                 shared_kwargs, accumulation_steps,
                 # Model architecture params
                 image_size, base_filters, depth_layer,
                 dropout, fundus_output_channels,
                 octa_output_channels, num_fundus_classes,
                 num_octa_classes):

        self.client_id = client_id

        # ── Store only plain Python objects — safe for Ray to pickle ──────────
        # File index lists contain Path objects and strings — pickleable
        self.octa_train_index       = octa_train_index
        self.octa_label_dict_train  = octa_label_dict_train
        self.fundus_idx_train       = fundus_idx_train
        self.fives_label_dict_train = fives_label_dict_train
        self.shared_kwargs          = shared_kwargs
        self.accumulation_steps     = accumulation_steps
        self.checkpoint_dir         = checkpoint_dir
        self.model_name             = model_name

        # Model architecture params
        self.image_size             = image_size
        self.base_filters           = base_filters
        self.depth_layer            = depth_layer
        self.dropout                = dropout
        self.fundus_output_channels = fundus_output_channels
        self.octa_output_channels   = octa_output_channels
        self.num_fundus_classes     = num_fundus_classes
        self.num_octa_classes       = num_octa_classes
        # ── TensorFlow objects built lazily in _setup() ───────────────────────
        # NOT set here — Ray serializes __init__ state, not _setup() state
        self._is_setup = False

    def _setup(self):
        """
        Build all TensorFlow objects the first time they are needed.
        Called at the start of fit() and get_parameters().
        Ray has already spawned the worker by the time this runs,
        so TensorFlow objects are created inside the worker process —
        never serialized across process boundaries.
        """
        if self._is_setup:
            return

        # Build model
        self.model = create_dual_modality_model(
            img_shape=(*self.image_size, 3),
            depth_layer=self.depth_layer,
            f=self.base_filters,
            dropout=self.dropout,
            AB_block=True,
            fundus_output_channels=self.fundus_output_channels,
            octa_output_channels=self.octa_output_channels,
            num_fundus_classes=self.num_fundus_classes,
            num_octa_classes=self.num_octa_classes,
            use_custom_training=True)

        self.model.compile(
            optimizer=keras.optimizers.Adam(learning_rate=1e-4),
            seg_weight=1.0,
            fundus_cls_weight=0.0,
            octa_cls_weight=0.0)
        print(f"THE TYPE OF self.octa_label_dict_train IS {self.octa_label_dict_train}")
        # Build datasets
        train_ds, train_samples = build_balanced_tf_dataset(
            self.fundus_idx_train, self.octa_train_index,
            fundus_label_dict=self.fives_label_dict_train,
            octa_label_dict=self.octa_label_dict_train,
            shuffle=True, **self.shared_kwargs)

        self.train_iter     = iter(train_ds)
        self.train_ds       = train_ds
        self.best_metric    = {}
        self.epochs_no_impr = {}
        self.epoch          = 0
        self.steps_per_epoch = int(np.ceil(train_samples / self.shared_kwargs['batch_size']))
        self._is_setup = True

    def get_parameters(self, config):
        self._setup()
        return self.model.get_weights()

    def fit(self, parameters, config):
        self._setup()
        self.model.set_weights(parameters)
        train_metrics = _train_one_epoch(
            self.model, self.train_iter, self.train_ds,
            self.steps_per_epoch, self.accumulation_steps)
        return self.model.get_weights(), len(self.octa_train_index), train_metrics
```

```python
def get_evaluate_fn(model, callbacks):
    """
    Returns a closure that Flower calls once per round on the server,
    after FedAvg has aggregated the client weights.

    Parameters
    ----------
    model     : the server-side model instance (same architecture as clients)
    callbacks : list of callbacks — ModelCheckpoint, EarlyStopping etc.
                These fire here, not on the clients.

    Returns
    -------
    evaluate_fn : callable with signature
                  (server_round, parameters, config) -> (loss, metrics)
                  which is exactly what Flower's strategy expects.
    """
    # Attach model to callbacks once
    for cb in callbacks:
        cb.set_model(model)
    best_metric    = {}
    epochs_no_impr = {}

    def evaluate_fn(server_round, parameters, config):
        # 1. Load the aggregated global weights into the server model
        model.set_weights(parameters)

        # 2. Run validation using our existing _evaluate function.
        #    epoch = server_round - 1 so epoch counting starts at 0.
        val_logs, stop_training = _evaluate(
            model,
            fundus_val_ds,  fundus_val_steps,
            octa_val_ds,    octa_val_steps,
            fundus_val_n,       octa_val_n,
            callbacks,
            epoch          = server_round - 1,
            best_metric    = best_metric,
            epochs_no_impr = epochs_no_impr)

        print(f"\n[Server] Round {server_round} evaluation:")
        for k, v in val_logs.items():
            print(f"  {k}: {v:.4f}")

        # 3. Flower requires (loss, metrics_dict)
        loss = float(val_logs['val_loss'])
        return loss, {k: float(v) for k, v in val_logs.items()}

    return evaluate_fn

def weighted_average_metrics(metrics):
    """
    Aggregate train metrics from all clients using sample-count weighting.
    `metrics` is a list of (num_samples, metrics_dict) tuples —
    one per client that returned from fit().
    """
    total_samples = sum(n for n, _ in metrics)

    aggregated = {}
    for num_samples, client_metrics in metrics:
        weight = num_samples / total_samples
        for key, value in client_metrics.items():
            # Only aggregate numeric scalars — skip anything else
            if isinstance(value, (int, float)):
                aggregated[key] = aggregated.get(key, 0.0) + float(value) * weight

    return aggregated
```
